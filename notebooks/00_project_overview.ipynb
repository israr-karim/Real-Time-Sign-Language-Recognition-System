# Project overview

This walkthrough uses the standard library and path configuration only. It does not load research models, download data, or create model outputs.

The three experiment notebooks are reorganized historical sources. Their model-quality results remain unverified.

```python
import json
import sys
from pathlib import Path

repository_root = next(
    (
        candidate
        for candidate in (Path.cwd(), *Path.cwd().parents)
        if (candidate / "src" / "sign_language_lab").is_dir()
    ),
    None,
)
if repository_root is None:
    raise RuntimeError("Open this notebook from inside the repository.")
source_directory = str(repository_root / "src")
if source_directory not in sys.path:
    sys.path.insert(0, source_directory)
```

## Original-to-recreated notebook index

```python
manifest = json.loads((repository_root / "docs" / "source_manifest.json").read_text())
for notebook in manifest["notebooks"]:
    print(notebook["original_filename"])
    print("  ->", notebook["recreated_file"])
```

## Recorded observations

These are saved historical observations with known validity problems, not a new benchmark. WER and BLEU are shown even when unfavorable. Cosine similarity is supplementary and is not an accuracy percentage.

```python
observations = json.loads(
    (repository_root / "results" / "legacy_observations.json").read_text()
)
print("Status:", observations["status"])
for run in observations["runs"]:
    print(run)
```

## Portable configuration

Each experiment has its own checkpoint location. Path resolution does not assert that the dataset or checkpoints are present.

```python
from sign_language_lab.config import ExperimentPaths

experiment_names = (
    "vivit_translation",
    "vivit_gloss_recognition",
    "videomae_experiment",
)
configurations = [ExperimentPaths.from_environment(name) for name in experiment_names]
assert len({configuration.checkpoint_file for configuration in configurations}) == 3
print("Three experiments have distinct checkpoint paths.")
```
